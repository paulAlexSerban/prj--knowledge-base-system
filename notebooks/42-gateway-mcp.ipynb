# MCP tools

The MCP server exposes the same four tools as the REST API. The client sends
the bearer token on the streamable HTTP transport. The two personas below
get the same sets as REST.


```python
import sys
from pathlib import Path

here = Path.cwd()
if not (here / "gateway_demo.py").exists():
    here = here / "notebooks"
if str(here) not in sys.path:
    sys.path.insert(0, str(here))

import gateway_demo
import hr_demo
```

```python
employee = gateway_demo.token("employee_eng")
outsider = gateway_demo.token("outsider")
names = gateway_demo.mcp_tools(employee)
assert {"search_knowledge", "get_document", "find_related", "list_sources"} <= set(names)

via_mcp = gateway_demo.mcp_call(
    employee, "search_knowledge", {"query": "parental leave Germany", "limit": 5}
)
via_rest = gateway_demo.search(employee, "parental leave Germany")
mcp_ids = {item["object_id"] for item in via_mcp["results"]}
rest_ids = {item["object_id"] for item in via_rest["results"]}
assert "pol-parental-leave-de" in mcp_ids
assert mcp_ids == rest_ids

outsider_mcp = gateway_demo.mcp_call(
    outsider, "search_knowledge", {"query": "parental leave Germany"}
)
assert outsider_mcp["results"] == []
print("mcp matches rest", sorted(mcp_ids))
```
