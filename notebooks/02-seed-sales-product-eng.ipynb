# Seed sales, product, and engineering

These three sources are a curated set. There is no bulk seed and no generated
people. `POST /admin/reset` puts the harness back to that set: 40 deals,
25 epics, and 12 specs.

Requires `make up`. The connectors pick the reset up on the next poll, or, for
product, from the webhook and the reconcile sweep.


```python
import sys
from pathlib import Path

here = Path.cwd()
if not (here / "source_demo.py").exists():
    here = here / "notebooks"
if str(here) not in sys.path:
    sys.path.insert(0, str(here))

import source_demo
```

```python
for harness, expected in (
    (source_demo.sales, {"deal": 40}),
    (source_demo.engineering, {"epic": 25}),
    (source_demo.product, {"spec": 12}),
):
    harness.reset()
    counts = harness.counts()
    print(harness.name, counts)
    if counts != expected:
        raise SystemExit(f"{harness.name} counts {counts}, expected {expected}")
```
