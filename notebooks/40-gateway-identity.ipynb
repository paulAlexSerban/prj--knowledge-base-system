# Gateway identity

Mint a token for each persona and check the cases that must fail.

`employee_eng` is an engineering employee. `hr_admin` is in People and can
read admin policies. `outsider` has no groups. The dev token endpoint is
`POST /dev/token`. A missing or bad bearer token is 401.


```python
import sys
from pathlib import Path

here = Path.cwd()
if not (here / "gateway_demo.py").exists():
    here = here / "notebooks"
if str(here) not in sys.path:
    sys.path.insert(0, str(here))

import gateway_demo
import hr_demo
```

```python
employee = gateway_demo.token("employee_eng")
admin = gateway_demo.token("hr_admin")
outsider = gateway_demo.token("outsider")
assert employee and admin and outsider and len({employee, admin, outsider}) == 3

missing = gateway_demo.raw("POST", "/v1/search", body={"query": "parental leave"})
assert missing.status_code == 401

bad = gateway_demo.raw(
    "POST", "/v1/search", token="not-a-token", body={"query": "parental leave"}
)
assert bad.status_code == 401

unknown = gateway_demo.raw("POST", "/dev/token", body={"persona": "no-such-persona"})
assert unknown.status_code == 404
print("personas ok")
```
